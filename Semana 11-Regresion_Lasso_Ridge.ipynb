# 🔍 **Regresión Lasso: Conceptos Clave**  

## 📌 **¿Qué es la Regresión Lasso?**  
La **Regresión Lasso** (Least Absolute Shrinkage and Selection Operator) es un método de regresión lineal que introduce una penalización en la magnitud de los coeficientes con el objetivo de reducir la complejidad del modelo y mejorar la capacidad de generalización.  

---

## 📊 **Fórmula Matemática**  
La regresión Lasso minimiza la siguiente función de costo:  

$
\mathcal{L}(\beta) = \sum_{i=1}^{n} (y_i - X_i \beta)^2 + \lambda \sum_{j=1}^{p} |\beta_j|
$

donde:  
- $ y_i $ son las observaciones de la variable dependiente.  
- $ X_i $ representa las variables independientes.  
- $ \beta_j $ son los coeficientes del modelo.  
- $ \lambda $ es el **parámetro de regularización**, que controla la penalización en los coeficientes.  

🛑 **Diferencia clave con Ridge Regression**: En lugar de penalizar los coeficientes con una norma $ L_2 $ ($ \sum \beta_j^2 $), Lasso usa la norma $ L_1 $, lo que tiende a forzar algunos coeficientes a ser exactamente cero, promoviendo la selección de variables.

---

## 🎯 **Ventajas y Desventajas**  

✅ **Ventajas:**  
✔️ Realiza selección de variables automáticamente.  
✔️ Reduce el sobreajuste al eliminar variables irrelevantes.  
✔️ Útil en datasets con muchas variables.  

⚠️ **Desventajas:**  
❌ Puede ser inestable cuando hay colinealidad alta.  
❌ No siempre es la mejor opción si todas las variables son importantes.  

---

## 🔧 **Cómo Elegir el Parámetro $ \lambda $**  
- Un $ \lambda $ **bajo** 🟢 da menos penalización y el modelo se comporta como una regresión lineal normal.  
- Un $ \lambda $ **alto** 🔴 fuerza muchos coeficientes a ser cero, eliminando variables.  
- Se puede seleccionar usando **validación cruzada**.  

---

La regresión Lasso tiene varios enfoques teóricos y perspectivas.   


## 🔍 **1. Interpretación como una Regularización $L_1$**
📌 **Concepto:**  
Lasso usa la penalización $L_1$, lo que tiende a hacer que algunos coeficientes sean exactamente cero, seleccionando automáticamente las variables más importantes.

📊 **Comparación con Ridge Regression:**  
- **Lasso:** Penaliza con $ L_1 $ ($ |\beta| $), lo que puede llevar a coeficientes exactamente **cero**.  
- **Ridge:** Penaliza con $ L_2 $ ($ \beta^2 $), lo que reduce los coeficientes pero sin anularlos completamente.  

📈 **Aplicación:** Útil cuando queremos hacer selección de características automáticamente.

---

## 🔢 **2. Interpretación en Términos de Optimización Convexa**
📌 **Concepto:**  
Lasso resuelve el siguiente problema de optimización convexa:

$
\min_{\beta} \sum_{i=1}^{n} (y_i - X_i \beta)^2 + \lambda \sum_{j=1}^{p} |\beta_j|
$

📊 **Geometría de la solución:**  
- La penalización $ L_1 $ genera un **espacio de restricciones en forma de diamante** en lugar de una esfera (como en Ridge).
- Esto provoca que la solución óptima se encuentre en los **vértices del diamante**, lo que explica por qué algunos coeficientes se anulan completamente.

🎯 **Conclusión:**  
Lasso **prefiere soluciones dispersas (sparse)**, eliminando coeficientes innecesarios.

---

## 🤖 **3. Relación con Métodos de Machine Learning**
📌 **Concepto:**  
Lasso se puede interpretar como un modelo de **aprendizaje supervisado** con regularización para evitar sobreajuste.

🔹 **Relación con árboles de decisión**: Ambos pueden eliminar características irrelevantes.  
🔹 **Relación con redes neuronales**: Se puede extender con regularización $ L_1 $ en capas ocultas.  

📈 **Aplicación:**  
- En modelos de regresión con **muchas variables** (Big Data).  
- Cuando queremos hacer **selección automática de características**.  

---

## 🔬 **4. Interpretación Bayesiana**
📌 **Concepto:**  
Lasso se puede ver como una estimación bayesiana con un **prior Laplaciano** sobre los coeficientes.

🔢 **Explicación matemática:**  
- En una regresión lineal clásica, asumimos un prior **Gaussiano** sobre $ \beta $, lo que equivale a Ridge Regression.  
- En Lasso, usamos un prior **Laplaciano**, que induce sparsity (coeficientes exactos en cero).  

📈 **Aplicación:**  
- Útil en modelos bayesianos cuando se requiere una selección natural de características.  
- Conexión con métodos de inferencia como MCMC.

La regresión Lasso utiliza varias métricas para evaluar su rendimiento. Aquí están las más comunes:

### 📊 **1. Error Cuadrático Medio (MSE)**
Mide el promedio de los errores al cuadrado entre las predicciones y los valores reales.

$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$

📌 **Más bajo es mejor**. Indica qué tan cerca están las predicciones de los valores reales.

---

### 📈 **2. Raíz del Error Cuadrático Medio (RMSE)**
Es la raíz cuadrada del MSE y se interpreta en la misma escala que la variable objetivo.

$
RMSE = \sqrt{MSE}
$

📌 **Más bajo es mejor**. Fácil de interpretar ya que está en las mismas unidades que $ y $.

---

### 🔍 **3. Error Absoluto Medio (MAE)**
Mide la diferencia promedio en valor absoluto entre las predicciones y los valores reales.

$
MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$

📌 **Más bajo es mejor**. Es más robusto ante valores atípicos que el MSE.

---

### ⚡ **4. Coeficiente de Determinación ($ R^2 $)**
Mide qué proporción de la variabilidad en la variable objetivo es explicada por el modelo.

$
R^2 = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}
$

📌 **Cercano a 1 es mejor**. Un $ R^2 $ cercano a 0 indica que el modelo no explica bien la variabilidad de los datos.

---

### 🔧 **5. Número de Coeficientes No Nulos**
Dado que Lasso anula coeficientes para seleccionar características, el número de coeficientes diferentes de cero es una métrica relevante.

📌 **Menos coeficientes no nulos** → Más regularización y mayor selección de variables.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Cargar el dataset de diamantes desde seaborn
sns.set(style="whitegrid")
diamonds = sns.load_dataset("diamonds")
````

````python
# Seleccionar las características relevantes para el modelo
features = ["carat", "depth", "table", "x", "y", "z", "cut", "color", "clarity"]
target = "price"
df = diamonds[features + [target]].copy()
````

````python
# Separar las características en numéricas y categóricas
categorical_features = ["cut", "color", "clarity"]
numeric_features = ["carat", "depth", "table", "x", "y", "z"]
````

````python
# Crear un preprocesador para escalar las variables numéricas y codificar las categóricas
preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_features),  # Normaliza las variables numéricas
    ("cat", OneHotEncoder(drop='first'), categorical_features)  # Convierte categóricas a variables dummy
])
````

````python
# Definir el parámetro de regularización (alpha)
# Se puede ajustar con validación cruzada para optimizar el rendimiento
alpha = 0.1  # Controla la penalización sobre los coeficientes

# Crear un pipeline que primero preprocesa los datos y luego aplica regresión Lasso
lasso = Pipeline([
    ("preprocessor", preprocessor),
    ("lasso", Lasso(alpha=alpha))
])
````

El codigo anterior, realiza la siguiente función

1. **Definir el parámetro de regularización (`alpha`)**  
   - `alpha = 0.1` establece la penalización en los coeficientes del modelo Lasso.  
   - Un valor mayor de `alpha` fuerza más coeficientes a ser cero (mayor regularización), mientras que un valor menor permite que más coeficientes tengan valores distintos de cero.

2. **Construir un `Pipeline` con dos pasos:**  
   - **`preprocessor`**: Aplica preprocesamiento a los datos, escalando variables numéricas y convirtiendo categóricas en variables dummy.  
   - **`lasso`**: Aplica la regresión Lasso con el valor de `alpha` previamente definido.

El uso del `Pipeline` permite que el preprocesamiento y la regresión se apliquen de manera secuencial y consistente.

````python
# Separar las variables predictoras (X) y la variable objetivo (y)
X = df[features]
y = df[target]

# Entrenar el modelo Lasso con los datos
lasso.fit(X, y)

# Obtener los coeficientes después del ajuste
feature_names = numeric_features + list(lasso.named_steps["preprocessor"].named_transformers_["cat"].get_feature_names_out())
coef = pd.Series(lasso.named_steps["lasso"].coef_, index=feature_names)
print(coef)
````

````python
# Graficar los coeficientes de la regresión Lasso
plt.figure(figsize=(10, 5))
coef.sort_values().plot(kind="barh")
plt.title(f"Coeficientes de la Regresión Lasso (alpha={alpha})")
plt.xlabel("Valor del coeficiente")
plt.ylabel("Características")
plt.show()
````

#🔵 **Regresión Ridge**  

## 📌 **¿Qué es la Regresión Ridge?**  
La **Regresión Ridge** es una variante de la regresión lineal que introduce **regularización** para evitar el sobreajuste y mejorar la generalización del modelo. Es un tipo de regresión lineal penalizada que **agrega una penalización L2** a la función de costo, lo que ayuda a reducir la varianza del modelo y controlar la magnitud de los coeficientes.  

## 🔍 **Función de Costo en Ridge**  
La regresión Ridge minimiza la siguiente función de costo:

\[
J(\beta) = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{p} \beta_j^2
\]

Donde:
- \( (y_i - \hat{y}_i)^2 \) es el **error cuadrático** de la predicción.  
- \( \lambda \sum_{j=1}^{p} \beta_j^2 \) es el **término de regularización**, que penaliza los coeficientes grandes.  
- \( \lambda \) (parámetro de regularización) controla el grado de penalización.  
  - Si \( \lambda = 0 \), se recupera la **regresión lineal estándar**.  
  - Si \( \lambda \) es muy grande, los coeficientes \( \beta_j \) se acercan a **cero**, reduciendo la complejidad del modelo.  

## ⚖️ **Diferencia entre Ridge y Lasso**  
| Característica        | Regresión Ridge | Regresión Lasso |
|----------------------|---------------|---------------|
| Penalización        | \( L2: \sum \beta_j^2 \) | \( L1: \sum |\beta_j| \) |
| Impacto en los coeficientes | Reduce su magnitud | Puede hacerlos exactamente cero |
| Uso en modelos con muchas variables | Mantiene todas las variables pero reduce su efecto | Hace selección de variables eliminando algunas |
| Mejor en...         | Datos con colinealidad | Datos con muchas variables irrelevantes |

## ✅ **Ventajas de Ridge**  
✔️ **Evita el sobreajuste**, especialmente cuando hay colinealidad entre variables.  
✔️ **Mantiene todas las variables** en el modelo, útil cuando se cree que todas tienen cierta influencia.  
✔️ **Controla la magnitud de los coeficientes**, reduciendo la sensibilidad a datos ruidosos.  

## ❌ **Desventajas de Ridge**  
❌ No realiza selección de variables, por lo que no es útil si se desea eliminar variables irrelevantes.  
❌ Depende de la correcta elección del **parámetro de regularización \( \lambda \)**, lo cual requiere validación cruzada.  

### 📈 **Ejemplo de Aplicación**  
La regresión Ridge es útil en problemas como:
- **Predicción de precios de casas** 🏠, donde muchas características pueden estar correlacionadas.  
- **Análisis de datos genéticos** 🧬, donde hay muchas variables altamente correlacionadas.  
- **Predicción de ventas** 📊, donde se manejan múltiples factores como precio, publicidad y promociones.  


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Cargar el dataset de diamantes desde seaborn
sns.set(style="whitegrid")
diamonds = sns.load_dataset("diamonds")
````

````python
# Seleccionar las características relevantes para el modelo
features = ["carat", "depth", "table", "x", "y", "z", "cut", "color", "clarity"]
target = "price"
df = diamonds[features + [target]].copy()
````

````python
# Separar las características en numéricas y categóricas
categorical_features = ["cut", "color", "clarity"]
numeric_features = ["carat", "depth", "table", "x", "y", "z"]
````

````python
# Crear un preprocesador para escalar las variables numéricas y codificar las categóricas
preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_features),  # Normaliza las variables numéricas
    ("cat", OneHotEncoder(drop='first'), categorical_features)  # Convierte categóricas a variables dummy
])

# Definir un valor fijo de alpha
alpha = 0.1  # Se puede ajustar manualmente según necesidad

# Crear un pipeline con Ridge
ridge = Pipeline([
    ("preprocessor", preprocessor),
    ("ridge", Ridge(alpha=alpha))
])
````

````python
# Separar las variables predictoras (X) y la variable objetivo (y)
X = df[features]
y = df[target]

# Entrenar el modelo Ridge
ridge.fit(X, y)
````

````python
# Obtener los coeficientes después del ajuste
feature_names = numeric_features + list(ridge.named_steps["preprocessor"].named_transformers_["cat"].get_feature_names_out())
coef = pd.Series(ridge.named_steps["ridge"].coef_, index=feature_names)
print(coef)
````

````python
# Graficar los coeficientes de la regresión Ridge
plt.figure(figsize=(10, 5))
coef.sort_values().plot(kind="barh")
plt.title(f"Coeficientes de la Regresión Ridge (alpha={alpha})")
plt.xlabel("Valor del coeficiente")
plt.ylabel("Características")
plt.show()
````

````python
# Evaluación del modelo con métricas de desempeño
y_pred = ridge.predict(X)

MSE = mean_squared_error(y, y_pred)  # Error cuadrático medio
RMSE = np.sqrt(MSE)  # Raíz del error cuadrático medio
MAE = mean_absolute_error(y, y_pred)  # Error absoluto medio
R2 = r2_score(y, y_pred)  # Coeficiente de determinación
n_nonzero_coef = np.sum(coef != 0)  # Número de coeficientes distintos de cero

# Imprimir los resultados de las métricas
print("Evaluación del modelo Ridge:")
print(f"MSE: {MSE:.2f}")
print(f"RMSE: {RMSE:.2f}")
print(f"MAE: {MAE:.2f}")
print(f"R^2: {R2:.4f}")
print(f"Número de coeficientes no nulos: {n_nonzero_coef}")
````

````python
# Gráfico de dispersión: Precio real vs. Predicho
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y, y=y_pred, alpha=0.5)
plt.plot([y.min(), y.max()], [y.min(), y.max()], '--r', linewidth=2)
plt.xlabel("Precio real")
plt.ylabel("Precio predicho")
plt.title("Precio real vs. Predicho")
plt.show()
````

````python
# Histograma de residuos
residuals = y - y_pred
plt.figure(figsize=(8, 6))
sns.histplot(residuals, bins=50, kde=True)
plt.xlabel("Error (y - y_pred)")
plt.ylabel("Frecuencia")
plt.title("Distribución de los residuos")
plt.show()
````

````python
# Gráfico de dispersión de residuos
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_pred, y=residuals, alpha=0.5)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel("Precio predicho")
plt.ylabel("Residuos")
plt.title("Gráfico de dispersión de residuos")
plt.show()
````
