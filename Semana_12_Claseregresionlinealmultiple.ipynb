## 📗 Introducción a la Regresión Lineal Múltiple

### 💡 ¿Qué es la Regresión Lineal Múltiple?
La regresión lineal múltiple es un método estadístico que permite modelar la relación entre una variable dependiente y dos o más variables independientes. A diferencia de la regresión lineal simple, donde se considera una única variable explicativa, en la regresión lineal múltiple se estudian múltiples factores que pueden influir en el valor de la variable objetivo.

La ecuación general de la regresión lineal múltiple es:
$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \ldots + \beta_n x_n + \epsilon$

Donde:
- **$y$**: variable dependiente (o variable objetivo).
- **$x_{\imath}$**: variables independientes (o predictoras).
- **$ \beta_0 $**: intercepto o término constante del modelo.
- **$ \beta_1, \beta_2, \ldots, \beta_n $**: coeficientes que representan la influencia de cada variable independiente sobre la variable dependiente.
- **$ \epsilon $**: término de error aleatorio, que captura factores no considerados en el modelo.

### 🔧 Suposiciones del Modelo
Para garantizar que el modelo sea válido y sus predicciones sean confiables, deben cumplirse ciertas suposiciones, entre ellas:
1. **Linealidad**: La relación entre las variables independientes y la variable dependiente debe ser aproximadamente lineal.
2. **No multicolinealidad**: Las variables independientes no deben estar fuertemente correlacionadas entre sí.
3. **Homoscedasticidad**: La varianza de los errores debe ser constante.
4. **Normalidad de los errores**: Los errores deben seguir una distribución normal.

### 📚 Aplicaciones Comunes
La regresión lineal múltiple es ampliamente utilizada en distintas áreas, como:
- **Economía**: para analizar el impacto de variables macroeconómicas en el crecimiento del PIB.
- **Ingeniería**: para predecir el desempeño de sistemas complejos basados en múltiples parámetros.
- **Medicina**: para identificar factores que influyen en el riesgo de desarrollar una enfermedad.


## 🎯 Estimación de los Parámetros: Métodos Principales
Para ajustar un modelo de regresión lineal múltiple y estimar sus parámetros (coeficientes $ \beta_0, \beta_1, \ldots, \beta_n $), existen varios métodos, siendo los más comunes:

#### 1. **Método de Mínimos Cuadrados Ordinarios (MCO)**
Este es el método clásico y más utilizado en regresión lineal. Busca minimizar la suma de los cuadrados de las diferencias entre los valores observados y los valores predichos por el modelo. La solución matemática se expresa como:
$ \hat{\beta} = (X^T X)^{-1} X^T y$

Donde:
- **$ X $**: matriz de diseño que contiene las variables independientes.
- **$ y $**: vector de la variable dependiente.
- **$ \hat{\beta} $**: vector de coeficientes estimados.

#### 2. **Método de Máxima Verosimilitud (MLE)**
El método de máxima verosimilitud busca encontrar los parámetros que maximicen la función de verosimilitud, es decir, aquellos que hacen que los datos observados sean más probables. Aunque es conceptualmente más complejo que el MCO, en modelos lineales bajo ciertas condiciones (errores normales), produce resultados equivalentes.

#### 3. **Regresión Ridge (Regularización L2)**
Este método introduce un término de penalización que evita el sobreajuste cuando hay muchas variables independientes correlacionadas (multicolinealidad). La función objetivo es:
$ \min \left( \sum (y_i - \hat{y}_i)^2 + \lambda \sum \beta_j^2 \right)$
Donde $ \lambda $ es un hiperparámetro que controla la cantidad de regularización.

#### 4. **Regresión Lasso (Regularización L1)**
Similar a la regresión Ridge, pero utiliza una penalización basada en la norma L1. Este método puede reducir coeficientes a cero, lo que permite realizar selección de variables.

#### 5. **Regresión Elastic Net**
Combina las penalizaciones de Ridge y Lasso, proporcionando un balance entre ambas técnicas.

### 🔎 Comparación de los Métodos
Cada método tiene sus propias ventajas y limitaciones. El MCO es adecuado cuando no hay problemas de multicolinealidad, mientras que los métodos regularizados (Ridge, Lasso, Elastic Net) son más úteiles en presencia de muchas variables correlacionadas o datasets con dimensionalidad alta.

| **Método**                         | **Ventajas**                                                                 | **Desventajas**                                                              | **Aplicaciones Recomendadas**                             |
|------------------------------------|------------------------------------------------------------------------------|-----------------------------------------------------------------------------|----------------------------------------------------------|
| **Mínimos Cuadrados Ordinarios (MCO)** | - Sencillo de implementar.                                                   | - Sensible a multicolinealidad y sobreajuste si hay muchas variables.        | - Casos sin problemas de multicolinealidad.               |
|                                    | - Proporciona estimaciones no sesgadas.                                       | - No maneja adecuadamente datasets con alta dimensionalidad.                 | - Modelos donde las suposiciones clásicas se cumplen.      |
| **Máxima Verosimilitud (MLE)**     | - Teóricamente robusto bajo distribuciones conocidas.                         | - Computacionalmente más costoso que MCO.                                    | - Modelos avanzados o cuando se desean distribuciones específicas. |
|                                    | - Produce estimaciones equivalentes a MCO bajo normalidad de errores.         |                                                                             |                                                          |

|                                    | - Reduce sobreajuste en datasets con alta dimensionalidad.                     | - No reduce coeficientes a cero (no realiza selección de variables).          |                                                          |

|                                    | - Balancea selección de variables y estabilidad del modelo.                   |                                                                             |                                                          |

### 🔎 Conclusión
La elección del método adecuado depende del contexto del problema:
- **MCO** es ideal cuando las suposiciones estándar del modelo lineal se cumplen.
- **Ridge** es útil cuando hay multicolinealidad entre las variables predictoras.
- **Lasso** puede ser la mejor opción si es necesario realizar una selección automática de variables.
- **Elastic Net** es adecuado para problemas donde hay muchas variables correlacionadas y es importante equilibrar la selección y regularización.


### 📊 **¿Qué es el VIF y para qué sirve en modelos de regresión múltiple?**

El **VIF** (Variance Inflation Factor, o **Factor de Inflación de Varianza**) es una medida clave para detectar la **multicolinealidad** en modelos de **regresión múltiple**. La **multicolinealidad** ocurre cuando las variables predictoras están altamente correlacionadas entre sí, lo que puede afectar la precisión y estabilidad de los coeficientes en el modelo.

### 🧮 **¿Qué mide el VIF?**
El VIF mide cuánto se **infla la varianza** de un coeficiente debido a la correlación de esa variable con las otras variables predictoras en el modelo.

Fórmula del VIF para una variable $X_{\jmath}$:


$VIF_j = \frac{1}{1 - R_j^2}$

donde \( R_j^2 \) es el **coeficiente de determinación** obtenido al hacer una regresión de \( X_j \) sobre las demás variables.

### 📏 **Interpretación de los valores de VIF**
- **VIF = 1**: 🔵 No hay colinealidad. Esta es la situación ideal.
- **1 < VIF < 5**: 🟢 Correlación moderada entre las variables. Generalmente no hay problemas graves.
- **VIF > 5**: 🟠 Alta multicolinealidad. Indica que las variables están fuertemente correlacionadas y puede afectar la interpretación.
- **VIF > 10**: 🔴 Grave multicolinealidad. Se recomienda tomar medidas correctivas.

### 🛠 **¿Por qué controlar el VIF?**
La **multicolinealidad** alta puede hacer que los coeficientes de las variables sean **inestables**, lo que dificulta la interpretación y puede reducir la confiabilidad de los resultados. Además, puede producir grandes variaciones en los coeficientes ante pequeños cambios en los datos.

### ⚙️ **¿Cómo corregir la multicolinealidad?**
1. **Eliminar variables correlacionadas**: Si dos o más variables están muy correlacionadas, considera eliminar una de ellas.
2. **Crear variables combinadas**: Generar nuevas variables combinando las variables correlacionadas.
3. **Usar técnicas de reducción de dimensionalidad**: Como el **PCA** (Análisis de Componentes Principales) para reducir el número de variables manteniendo la mayor cantidad de información posible.



## 🏠 Cargar el Dataset:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

# 1. Cargar el dataset de California Housing
california = datasets.fetch_california_housing()

# Convertir a DataFrame para mayor legibilidad
df = pd.DataFrame(california.data, columns=california.feature_names)
df['PRICE'] = california.target  # Agregar la variable objetivo 'PRICE'


```

## 📊 Análisis Exploratorio de Datos (EDA):

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

# 2. Omitir las columnas 'Latitude' y 'Longitude' para el análisis
df = df.drop(columns=['Latitude', 'Longitude'])

# 3. Análisis exploratorio de los datos (EDA)
print("Primeras filas del dataset:")
print(df.head())

# Descripción estadística del dataset
print("\nDescripción estadística del dataset:")
print(df.describe())

# Comprobamos si hay valores nulos
print("\nComprobando valores nulos:")
print(df.isnull().sum())

# 4. Visualización de scatterplots entre todas las características y la variable objetivo 'PRICE'
plt.figure(figsize=(15, 10))
for i, column in enumerate(df.columns[:-1]):  # Excluimos la columna 'PRICE'
    plt.subplot(3, 3, i+1)
    sns.scatterplot(x=df[column], y=df['PRICE'], alpha=0.5)
    plt.title(f"Relación entre {column} y PRICE")
    plt.xlabel(column)
    plt.ylabel("Precio de la vivienda")

plt.tight_layout()
plt.show()

# 5. Heatmap de correlación entre todas las variables
plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', cbar=True, linewidths=0.5)
plt.title("Heatmap de Correlación entre Variables")
plt.show()

# 6. Multicolinealidad: Cálculo de VIF (Variance Inflation Factor)
# El VIF evalúa la colinealidad de cada variable con respecto a las demás. VIF > 5 o 10 indica posible multicolinealidad.

# Agregar constante para calcular el VIF
X = add_constant(df.drop(columns=['PRICE']))  # Matriz de características sin la variable dependiente
vif_data = pd.DataFrame()
vif_data['Variable'] = X.columns
vif_data['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

print("\n### Análisis de Multicolinealidad ###")
print(vif_data)

# 7. Visualización de VIF
plt.figure(figsize=(10, 6))
sns.barplot(x='VIF', y='Variable', data=vif_data, palette='coolwarm')
plt.axvline(x=5, color='red', linestyle='--', label='Límite de multicolinealidad (VIF > 5)')
plt.xlabel('VIF (Variance Inflation Factor)')
plt.title('Gráfico de VIF para Identificar Multicolinealidad')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
coefficients = model.coef_
intercept = model.intercept_

print(f"\nCoeficientes estimados: {coefficients}")
print(f"Término independiente estimado: {intercept}")
#######################################################
print('\n')
equation = f"PRICE = {intercept} "
for i, coef in enumerate(coefficients):
  equation += f"+ ({coef} * {X.columns[i]}) "

equation #se obtiene el modelo con los coeficientes
```

##📊 Interpretacion de los coeficientes

### 1. **Coeficientes Estimados:**

Los **coeficientes** son los valores que indican la relación entre cada característica (variable independiente) y la variable dependiente (en este caso, el **precio de la vivienda**). Cada coeficiente nos dice cuánto **cambia el precio de la vivienda** por cada unidad de cambio en la característica correspondiente, **manteniendo todas las demás características constantes**.

Lo que se está mostrando en tu salida son los **coeficientes estimados** y el **término independiente** de un modelo de regresión lineal múltiple.

En este caso, el modelo tiene la siguiente forma general:


$\text{PRICE} = \beta_0 + \beta_1 \cdot \text{MedInc} + \beta_2 \cdot \text{HouseAge} + \beta_3 \cdot \text{AveRooms} + \beta_4 \cdot \text{AveBedrms} + \beta_5 \cdot \text{Population} + \beta_6 \cdot \text{AveOccup}$

Donde:
- **PRICE** es la variable dependiente (el precio de la vivienda).
- **MedInc**, **HouseAge**, **AveRooms**, **AveBedrms**, **Population**, y **AveOccup** son las variables independientes (las características de las viviendas o el área).
- Los **coeficientes** (denotados como $\beta_0, \beta_1, \ldots, \beta_6$) indican la relación entre cada variable independiente y la variable dependiente.
