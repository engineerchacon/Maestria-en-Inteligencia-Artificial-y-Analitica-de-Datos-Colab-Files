# El Valor del Preprocesamiento

En la actualidad, **los datos no tienen valor por sí mismos**, sino por el conocimiento y la información que pueden generar. Tal como lo plantea el principio ampliamente conocido de "garbage in, garbage out", acumular grandes volúmenes de datos sin una adecuada preparación no garantiza resultados de calidad; al contrario, puede conducir a conclusiones erróneas y decisiones mal fundamentadas.

En este contexto, el concepto de **Smart Data (datos inteligentes)** surge como una evolución crítica. Hace referencia al desarrollo y aplicación de herramientas capaces de procesar datos masivos, heterogéneos y no estructurados, con el objetivo de revelar su verdadero valor y potencial analítico

---
Una vez que los **datos han sido transformados en información útil** —es decir, convertidos en Smart Data—, se **posibilitan interacciones eficientes** y en tiempo real con **aplicaciones de inteligencia empresarial, sistemas de soporte a decisiones y plataformas transaccionales**. Esto marca la transición de organizaciones centradas únicamente en la gestión de datos hacia organizaciones de aprendizaje, donde el conocimiento se convierte en el núcleo estratégico.


A medida que los **conjuntos de datos crecen **—particularmente en cuanto a su dimensionalidad y número de ejemplos—, **también lo hace la presencia de ruido**, lo cual **compromete la calidad de los datos y afecta directamente el rendimiento de los modelos de aprendizaje automático**. Este fenómeno ha sido ampliamente documentado: a mayor dimensionalidad/instancias, mayor riesgo de acumulación de ruido, lo que puede derivar en inestabilidad algorítmica, es decir, fluctuaciones significativas en el rendimiento de los modelos ante pequeños cambios en los datos.

En este contexto, para que los datos puedan considerarse verdaderamente inteligentes, es indispensable someterlos a procesos de limpieza y depuración, incluso después de haber sido integrados o transformados.

---
Como ya se ha abordado en el tema de **Reducción de la Dimensionalidad**, el objetivo general de la **reducción de datos** es obtener una representación más compacta del conjunto de datos original, que sea más manejable y eficiente para su análisis, sin perder la información relevante para la tarea predictiva o descriptiva.

La reducción se puede llevar a cabo eliminando o sustituyendo instancias (filas/ejemplos) del conjunto de datos. Esto recibe el nombre de **numerosity reducction**

### Necesitamos reducir los datos porque…
- El análisis de datos podría ser demasiado complejo con conjuntos de datos enormes.
- El análisis de datos podría ser demasiado costoso computacionalmente con conjuntos de datos enormes.
- Las instancias ruidosas pueden provocar clasificaciones erróneas.
- Los atributos irrelevantes y redundantes pueden obstaculizar cualquier proceso de aprendizaje automático.

#Reducción de la Numerosidad: Significado

Con el fin de **enfocar** el **análisis en los datos realmente valiosos**, las **técnicas de reducción de datos juegan un papel clave**. Estas buscan eliminar ejemplos redundantes, contradictorios o atípicos, lo cual no solo mejora la eficiencia computacional, sino que también contribuye a aumentar la precisión y estabilidad de los modelos predictivos.

La **reducción de la numerosidad** es una técnica de preprocesamiento de datos cuyo objetivo principal es **disminuir la cantidad de ejemplos** (instancias o filas) en un conjunto de datos, **sin comprometer** su **capacidad **para representar adecuadamente el problema de análisis o predicción.

A diferencia de la reducción de la dimensionalidad —que busca disminuir el número de atributos o características (columnas)—, la **reducción de la numerosidad actúa sobre la escala horizontal del conjunto de datos (las observaciones)**.





# Reducción de la Numerosidad: Origen

El algoritmo de los k vecinos más cercanos (kNN es uno de los clasificadores no paramétricos más conocidos y ampliamente utilizados en tareas de aprendizaje automático y minería de datos. A pesar de su simplicidad conceptual y eficacia en muchos escenarios, es bien sabido que kNN presenta varias desventajas operativas. En particular, existen tres limitaciones clave que afectan su aplicación exitosa:

- Elevados requerimientos de almacenamiento: kNN necesita conservar la totalidad del conjunto de entrenamiento en memoria, ya que su regla de decisión depende directamente de los ejemplos almacenados.

- Baja eficiencia computacional: durante la fase de predicción, el algoritmo debe calcular distancias entre cada muestra de prueba y todas las muestras de entrenamiento, lo que genera un alto costo computacional, especialmente en conjuntos grandes.

- Baja tolerancia al ruido: kNN —y en especial su variante 1NN— considera todos los ejemplos como igualmente válidos, lo que lo hace susceptible a errores cuando el conjunto de entrenamiento contiene instancias ruidosas, erróneas o atípicas.

---

Numerosas investigaciones han abordado estas limitaciones, proponiendo mejoras en dos frentes principales:

- Por un lado, se han propuesto nuevas métricas de similitud y funciones de distancia adaptadas a entornos de alta dimensionalidad, con el objetivo de mejorar la precisión del clasificador.

- Por otro, se han explorado diversas estrategias para reducir los costos computacionales, tales como la partición del espacio de características, el cálculo restringido a regiones vecinas o el uso de estructuras de indexación eficientes, como árboles k-d o árboles R.

---

Una de las aproximaciones más efectivas para abordar simultáneamente los tres problemas mencionados —complejidad, almacenamiento y sensibilidad al ruido— es la reducción de datos mediante selección de instancias, ejemplos, o prototipos. Estos últimos tres términos son sinónimos.



# Reducción de la Numerosidad: Taxonomía

## Selección de Instancias

Las técnicas de selección de instancias buscan generar un subconjunto del conjunto de entrenamiento original que sea representativo, compacto y eficaz, es decir, que mantenga o incluso mejore el rendimiento de clasificación del modelo sobre datos nuevos, al tiempo que reduce el volumen de datos requeridos.

Estas técnicas son conocidas en la literatura con distintos términos, tales como: Técnicas de reducción de datos, Selección de instancias, o Selección de prototipos (PS).

---
### Tipos de Selección de Instancias

La selección de instancias puede clasificarse en dos enfoques principales, dependiendo del objetivo que se persiga:

- **Limpieza de datos** (*data cleaning*)
  - Se enfoca en **mejorar el rendimiento del modelo eliminando ejemplos ruidosos, inconsistentes o atípicos** que podrían degradar la calidad de la predicción.

- **Condensación de datos** (*data condensing*)
  - Busca **reducir el tamaño del conjunto de entrenamiento**, **conservando** únicamente aquellas **instancias que son esenciales para la clasificación**, lo que permite **disminuir los costos computacionales sin sacrificar (o incluso mejorando) la precisión**.

---
###¿Cómo se logran los objetivos de la selección de instancias?###

La selección de instancias se implementa a través de estrategias específicas que dependen del enfoque adoptado:

**Limpieza de Datos**: Identificación y manejo de instancias con ruido, es decir, ejemplos mal etiquetados o inconsistentes, los cuales pueden ser *modificados*, *reemplazados* o *eliminados*:

- *Tratamiento de valores faltantes en atributos clave (missing data)*, ya sea a través de imputación, eliminación o modificación.

- *Eliminación o corrección de ejemplos incorrectos o inexactos (noisy data)*, que pueden afectar negativamente el proceso de aprendizaje.

**Condensación de datos**: *Identificación de instancias redundantes* que no aportan nueva información, las cuales seràn eliminadas.




### ¿Qué son los datos ruidosos?

En el contexto del aprendizaje automático, los **datos ruidosos** son aquellos que contienen **errores, inconsistencias o información no confiable**, producto de fallos durante la **recolección, codificación o etiquetado de los datos**.

Existen principalmente **dos tipos de ruido**:

- **Ruido en atributos**: ocurre cuando los **valores de las variables** (características) de las muestras son incorrectos o imprecisos. Por ejemplo, una lectura sensorial con error o un campo numérico mal registrado.
  
- **Ruido en etiquetas (o ruido de clase)**: se refiere a casos en los que la **etiqueta de clase asignada a una instancia es incorrecta**. Este tipo de ruido puede ser más perjudicial, ya que afecta directamente la supervisión del modelo durante el entrenamiento.

### ¿Por qué ocurre el ruido en las etiquetas?

El ruido de clase puede originarse por diversas razones:

- Etiquetado subjetivo o inconsistente.
- Etiquetas asignadas por **expertos no especializados** o sin conocimiento suficiente.
- **Errores humanos** o fallos en sistemas automáticos de etiquetado.
- Problemas de **codificación**, **ambigüedad semántica** o mala comunicación durante el registro.

---

## ¿Cómo afecta el ruido de clase al rendimiento del modelo?

Si el ruido de clase **no se detecta ni se corrige adecuadamente**, puede tener consecuencias importantes en el proceso de aprendizaje:

-  **Disminución en el rendimiento predictivo**: los clasificadores aprenden patrones incorrectos, reduciendo su capacidad de generalización.
- **Alteración de los límites de decisión**: el modelo puede ajustar fronteras más complejas e imprecisas, debido a etiquetas contradictorias.
-  **Mayor solapamiento entre clases**: se incrementa la confusión entre categorías, especialmente si el ruido se distribuye de forma no aleatoria.
- **Sobreajuste (overfitting)**: el modelo intenta aprender ruidos como si fueran patrones reales, lo que afecta su rendimiento en datos nuevos.
- **Mayor complejidad del problema**: el proceso de entrenamiento se vuelve más difícil y la interpretación del modelo más ambigua.

---



### ¿Qué es un valor atípico?

Un **valor atípico** (o *outlier*) es una instancia que **se desvía significativamente del comportamiento general del resto del conjunto de datos**. Estos valores suelen ser casos **raros, anómalos o extremos**, y pueden surgir por diversas razones:

- Errores de medición o registro.
- Procesos distintos a los que predominan en los datos.
- Comportamientos legítimos pero poco frecuentes (*eventos excepcionales*).

Los valores atípicos son más comunes en conjuntos de datos grandes, y aunque su **presencia aislada no siempre afecta de forma significativa al análisis**, pueden influir en modelos sensibles a la distancia (como k-NN o regresión), alterar estadísticas agregadas (como la media), y dificultar la identificación de patrones regulares.

---

## Diferencia entre valores atípicos y ruido de clase

| Concepto            | Valor atípico                                        | Ruido de clase                                       |
|---------------------|------------------------------------------------------|------------------------------------------------------|
| **Naturaleza**       | Instancia con atributos extremos o inusuales         | Instancia con **etiqueta de clase incorrecta**       |
| **Origen posible**   | Error de medición o **evento raro pero válido**      | Error de codificación, etiquetado o ambigüedad       |
| **Puede ser útil?**  | Sí, puede revelar patrones raros e interesantes      | No, **generalmente distorsiona** el aprendizaje      |
| **Impacto esperado** | Influye en la distribución o en la frontera del modelo | Puede generar **solapamiento artificial** y ruido    |

> 🔎 **Importante:** No todos los valores atípicos son errores. Algunos pueden representar eventos excepcionales y merecen ser analizados antes de decidir eliminarlos.

### Visualización de Tipos de Instancias en Datos Sintéticos

En la gráfica se utilizan colores y formas de marcadores para diferenciar tanto las dos clases como los tipos de instancias representados:

- Colores (p. ej., azul vs. naranja): distinguen las dos clases de datos sintéticos. Cada clase tiene un color asignado, de modo que los grupos de puntos de cada color corresponden a cada categoría/clase original.
- Círculo – Instancia original: punto de datos correctamente etiquetado, sin contaminación de ruido. Representa un ejemplo típico y confiable de su clase.
- Triángulo – Instancia con etiqueta ruidosa: indica un ejemplo cuya etiqueta de clase es incorrecta (ruido en la etiqueta). Estos puntos suelen aparecer en medio del cúmulo de la otra clase (color opuesto), resaltando que están mal asignados con respecto a su verdadera clase.
- Cuadrado – Instancia con atributos ruidosos: representa una instancia cuyo vector de características ha sido alterado por ruido. Su posición en el plano puede ser atípica (outlier) respecto al resto de su clase de color, mostrando que sus atributos no son fiables y la ubican lejos del patrón esperado.
- Cuadrado – Instancia redundante: un ejemplo duplicado o muy similar a otro. Suele aparecer superpuesto o muy cercano a otro punto de la misma clase. Aunque correctamente etiquetado, aporta información repetida que no añade valor nuevo debido a su redundancia.
- Estrella – Instancia irrelevante: un punto de datos que resulta ajeno a los patrones de las dos clases principales. Su color puede coincidir con alguna clase o ser neutro, pero conceptualmente es un outlier extremo o dato espurio que no contribuye a definir la frontera entre clases (es irrelevante para el aprendizaje del modelo).


```python
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Generar datos sintéticos
X, y = make_classification(n_samples=300, n_features=2, n_informative=2,
                           n_redundant=0, n_clusters_per_class=1,
                           class_sep=1.5, flip_y=0, random_state=42)
df = pd.DataFrame(X, columns=['Atributo_1', 'Atributo_2'])
df['Clase'] = y
df['Tipo'] = 'Original'

# Etiquetas ruidosas
np.random.seed(42)
idx_ruido_etiqueta = np.random.choice(df.index, size=int(0.1 * len(df)), replace=False)
df.loc[idx_ruido_etiqueta, 'Clase'] = 1 - df.loc[idx_ruido_etiqueta, 'Clase']
df.loc[idx_ruido_etiqueta, 'Tipo'] = 'Etiqueta ruidosa'

# Atributos ruidosos
idx_ruido_atributo = np.random.choice(df.index, size=int(0.1 * len(df)), replace=False)
df.loc[idx_ruido_atributo, 'Atributo_1'] += np.random.normal(0, 3, size=len(idx_ruido_atributo))
df.loc[idx_ruido_atributo, 'Tipo'] = 'Atributo ruidoso'

# Redundantes (duplicados)
idx_duplicados = np.random.choice(df.index, size=10, replace=False)
df_duplicados = df.loc[idx_duplicados].copy()
df_duplicados['Tipo'] = 'Redundante'
df = pd.concat([df, df_duplicados], ignore_index=True)

# Irrelevantes (aleatorios)
irrelevantes = pd.DataFrame({
    'Atributo_1': np.random.uniform(-5, 5, 10),
    'Atributo_2': np.random.uniform(-5, 5, 10),
    'Clase': np.random.randint(0, 2, 10),
    'Tipo': 'Irrelevante'
})
df = pd.concat([df, irrelevantes], ignore_index=True)

# Estilo por clase y tipo
colores_clase = {0: '#1f77b4', 1: '#ff7f0e'}
df['Etiqueta'] = df['Tipo'] + ' - Clase ' + df['Clase'].astype(str)
etiquetas_unicas = df['Etiqueta'].unique()

marcadores = {
    'Original': 'o',
    'Etiqueta ruidosa': 'x',
    'Atributo ruidoso': '^',
    'Redundante': 's',
    'Irrelevante': '*'
}

estilos = {}
for etiqueta in etiquetas_unicas:
    tipo, clase = etiqueta.split(' - Clase ')
    clase = int(clase)
    estilos[etiqueta] = {
        'color': colores_clase[clase],
        'marker': marcadores.get(tipo, 'o')
    }

# Visualización sin warnings
plt.figure(figsize=(12, 7))
for etiqueta, estilo in estilos.items():
    subset = df[df['Etiqueta'] == etiqueta]
    if subset.empty:
        continue

    marker = estilo['marker']
    color = estilo['color']

    if marker in ['x', '+', '|', '_']:  # marcadores no rellenos
        plt.scatter(subset['Atributo_1'], subset['Atributo_2'],
                    c=color, marker=marker, label=etiqueta,
                    s=80, alpha=0.7)
    else:
        plt.scatter(subset['Atributo_1'], subset['Atributo_2'],
                    c=color, marker=marker, label=etiqueta,
                    edgecolors='k', s=80, alpha=0.7)

plt.title('Visualización sin warnings: ruido, redundancia e irrelevancia por clase')
plt.xlabel('Atributo 1')
plt.ylabel('Atributo 2')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Estrategias para gestionar el ruido en aprendizaje automático

La presencia de ruido —especialmente en las etiquetas de clase— puede comprometer seriamente el rendimiento de los modelos predictivos. Por ello, se han desarrollado **diversas estrategias para mitigar su impacto**, las cuales se agrupan en dos niveles principales:

---

### 1. Enfoques a nivel de algoritmo

Estos métodos buscan **construir clasificadores robustos o tolerantes al ruido**, sin necesidad de modificar previamente los datos. La idea es que el algoritmo pueda manejar por sí mismo la presencia de ruido durante el aprendizaje.

- **Ejemplos comunes:**
  - Árboles de decisión como **C4.5**.
  - Técnicas de **boosting**, que corrigen errores iterativamente.
  - Modelos regulares como **SVM con soft margin**.

>  **Ventaja**: no requiere preprocesamiento adicional.  
> **Limitación**: puede no ser suficiente si el nivel de ruido es alto o si se busca simplificar el conjunto de entrenamiento.

---

### 2. Enfoques a nivel de datos

Estos enfoques se aplican **antes de entrenar un modelo**, y buscan **detectar, eliminar o corregir instancias ruidosas**. Son especialmente útiles cuando no se dispone de algoritmos robustos, o se desea mejorar aún más su rendimiento.

#### Métodos de filtrado (o edición)

- **Objetivo**: eliminar instancias sospechosas de ser ruidosas (e.g., cercanas al límite de decisión).
- **Resultado esperado**: suavizar los límites entre clases y **reducir el tamaño del conjunto de datos**.
- **Técnica típica**: identificar ejemplos inconsistentes a través de reglas de vecindad (como k-NN inverso).

#### Métodos de reetiquetado

- **Objetivo**: modificar la etiqueta de clase de las instancias cuya clase parece incorrecta.
- **Método**: se asigna una nueva clase basada en la información local (vecindad, confianza, consenso entre clasificadores).
- **Advertencia**: si se reetiquetan ejemplos correctos por error, se puede introducir sesgo.

---

### Ejemplos de técnicas de preprocesamiento a nivel de datos

- **Edición múltiple (Repeated Edition)**
- **Filtro de partición iterativo**
- **Edición basada en grafos** (grafo de Gabriel, grafo RNG)
- **Selección difusa de prototipos aproximados**
- **Edición recompensa-castigo**
- **Selección de modelo de clase**
- **Filtrado iterativo de casos**
- **Función base de edición radial normalizada**

> *Estas técnicas suelen implementarse como parte de pipelines de selección de instancias o reducción de conjuntos de entrenamiento.*

---


### Condensación de datos

La **condensación de datos** es una técnica de reducción que busca generar un **subconjunto representativo del conjunto de entrenamiento original**, conservando su capacidad predictiva.

### Objetivo

- **Reducir el tamaño del conjunto de entrenamiento**.
- **Mantener (o incluso mejorar) el rendimiento del clasificador**.
- Disminuir los **requisitos de almacenamiento y cómputo**, haciendo el modelo más eficiente sin sacrificar precisión.

---

## ¿Cómo funcionan los algoritmos de condensación?

Los algoritmos de condensación aplican principios de **simplificación estructural** del conjunto de entrenamiento:

- Eliminan **instancias irrelevantes o redundantes** que no contribuyen significativamente a la discriminación entre clases.
- Usualmente remueven las **instancias internas** (es decir, aquellas alejadas de las regiones de decisión), ya que son seguras de eliminar sin afectar la clasificación.
- Mantienen preferentemente las **instancias cercanas a los límites de decisión**, que son más informativas para el modelo.

> Esto es opuesto a los métodos de filtrado, que tienden a eliminar las instancias ruidosas o mal etiquetadas, muchas veces ubicadas en los márgenes.

---

## Consistencia del conjunto de entrenamiento

Una propiedad clave en condensación es la **consistencia**:

> Dado un conjunto de entrenamiento no vacío $T$ $(T <> ∅)$, un subconjunto $S$ de $T$ ($S⊆ T$) es **consistente con respecto a $T$** si al usar $S$ como conjunto de entrenamiento, el clasificador **1-NN** logra clasificar **correctamente todas las instancias de \( T \)**.

*Es decir, S es consistente si 1-NN entrenado con S tiene 100 % de Accuracy sobre el conjunto original T.*

Esto permite reducir el conjunto de datos sin pérdida de información relevante para la tarea de clasificación.

## Prácticas con Edición de Wilson

## Edición de Wilson (Edited Nearest Neighbor - ENN)

La **edición de Wilson** es una técnica de limpieza de datos supervisada que busca **mejorar la calidad del conjunto de entrenamiento** mediante la eliminación de instancias que probablemente sean ruidosas o ambiguas.

### ¿Cómo funciona?

1. Para cada instancia en el conjunto de entrenamiento, se evalúa si su clase coincide con la mayoría de sus `k` vecinos más cercanos (usualmente `k=3`).
2. Si **la clase predicha por sus vecinos no coincide con su etiqueta real**, la instancia se considera **ruidosa** y se elimina.
3. El resultado es un conjunto más limpio, con **fronteras de decisión más suaves** y menor solapamiento entre clases.

### ¿Por qué usarlo?

- Elimina ruido de clase (etiquetas incorrectas).
- Suaviza los límites entre clases.
- Mejora el rendimiento de clasificadores sensibles a la estructura de los datos, como `k-NN`.

### Nota

La edición de Wilson **no busca reducir el conjunto al mínimo (como CNN)**, sino **mejorar su calidad eliminando ejemplos conflictivos**.

## Generación de datos sintéticos con clases concéntricas

Este ejemplo muestra cómo generar un conjunto de datos artificial con dos clases distribuidas de forma **concéntrica** (una dentro de otra), utilizando la función `make_circles` de Scikit-learn.

- Se generan 500 muestras con dos atributos.
- El parámetro `noise=0.2` introduce **variabilidad (ruido)** en las posiciones, generando cierto **solapamiento entre las clases**.
- Este tipo de datos es útil para ilustrar problemas **no linealmente separables** y evaluar el rendimiento de clasificadores como SVM con kernel o redes neuronales.

La visualización permite identificar claramente ambas clases y su distribución espacial.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
import pandas as pd

# Generar datos con dos clases concéntricas y algo de solapamiento
X, y = make_circles(n_samples=2500, factor=0.5, noise=0.2, random_state=42)

# Convertir a DataFrame para mayor claridad
df = pd.DataFrame(X, columns=['Atributo_1', 'Atributo_2'])
df['Clase'] = y

# Visualización
plt.figure(figsize=(8, 6))
for clase in [0, 1]:
    subset = df[df['Clase'] == clase]
    plt.scatter(subset['Atributo_1'], subset['Atributo_2'],
                label=f'Clase {clase}', alpha=0.7, s=80, edgecolor='k')

plt.title('Clases concéntricas con solapamiento')
plt.xlabel('Atributo 1')
plt.ylabel('Atributo 2')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from sklearn.utils.validation import check_X_y
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
import random

class ENN:
    def __init__(self, n_neighbors=3):
        self.n_neighbors = n_neighbors
        self.classifier = KNeighborsClassifier(n_neighbors=self.n_neighbors)
        self.sample_indices_ = []

    def fit(self, X, y):
        # Validación inicial
        X, y = check_X_y(X, y, accept_sparse=False)

        mask = np.ones(len(y), dtype=bool)
        tmp_m = np.ones(len(y), dtype=bool)

        indices = list(range(len(y)))
        random.shuffle(indices)

        for i in indices:
            tmp_m[i] = False
            self.classifier.fit(X[tmp_m], y[tmp_m])
            x_i = X[i].reshape(1, -1)
            y_pred = self.classifier.predict(x_i)
            if y_pred[0] != y[i]:
                mask[i] = False
            tmp_m[i] = True

        self.X_ = X[mask]
        self.y_ = y[mask]
        self.sample_indices_ = np.where(mask)[0]
        self.reduction_ = 1 - len(self.y_) / len(y)
        return self
```

## Analizando el Efecto de WIlson en los Clasificadores



```python
from sklearn.datasets import make_classification
from sklearn.model_selection import StratifiedKFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import numpy as np
import pandas as pd


# ---------- Clasificadores ----------
classifiers = {
    "KNN": KNeighborsClassifier(n_neighbors=3),
    "Decision Tree": DecisionTreeClassifier(random_state=0)
}

# ---------- Evaluación por fold con y sin ENN ----------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
results_by_fold = {name: {'with_ENN': [], 'without_ENN': []} for name in classifiers}

for train_idx, test_idx in cv.split(X, y):
    X_train, y_train = X[train_idx], y[train_idx]
    X_test, y_test = X[test_idx], y[test_idx]

    # ---- Sin ENN
    for name, clf in classifiers.items():
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        results_by_fold[name]['without_ENN'].append(acc)

    # ---- Con ENN
    enn = ENN(n_neighbors=3)
    enn.fit(X_train, y_train)
    X_train_clean = X_train[enn.sample_indices_]
    y_train_clean = y_train[enn.sample_indices_]

    for name, clf in classifiers.items():
        clf.fit(X_train_clean, y_train_clean)
        y_pred = clf.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        results_by_fold[name]['with_ENN'].append(acc)

# ---------- Mostrar resultados ----------
summary = pd.DataFrame({
    f"{name} - Sin ENN (Promedio)": f"{np.mean(results_by_fold[name]['without_ENN']):.4f}"
    for name in classifiers
}, index=["Accuracy promedio"])

for name in classifiers:
    summary[f"{name} - Con ENN (Promedio)"] = f"{np.mean(results_by_fold[name]['with_ENN']):.4f}"

print(summary)
```

## Práctica con Algoritmo de Condensanción: Condensado de Hart




##Condensado de Hart (Condensed Nearest Neighbor - CNN)

El algoritmo **Condensado de Hart**, también conocido como **CNN**, es una técnica de **reducción de la numerosidad** propuesta por *Peter E. Hart* en 1968.

### Objetivo

Reducir el conjunto de entrenamiento sin comprometer la capacidad de clasificación del modelo **1-NN**, manteniendo solo las instancias esenciales.

### ¿Cómo funciona?

1. Se inicializa el conjunto condensado con al menos una instancia por clase.
2. Se recorren las instancias restantes y:
   - Si una instancia **no puede ser clasificada correctamente** por el conjunto actual, se **añade** al condensado.
3. El proceso se repite hasta que todas las instancias estén bien clasificadas o ya no se añadan nuevas.

### Ventajas

- Reduce el tiempo de clasificación en k-NN.
- Conserva la estructura esencial del conjunto.
- Selecciona instancias representativas cercanas a la frontera de decisión.

### 📚 Referencia

> Hart, P. E. (1968). *The condensed nearest neighbor rule*. IEEE Transactions on Information Theory, 14(3), 515–516.



```python
from sklearn.utils.validation import check_X_y
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
import random

class CNN:
    """
    Condensed Nearest Neighbors (CNN) - versión adaptada para datos densos.

    Este algoritmo reduce el conjunto de entrenamiento reteniendo solo
    las instancias necesarias para preservar la capacidad de clasificación
    bajo la regla 1-NN (o k-NN si se ajusta n_neighbors).

    Basado en:
    P. E. Hart, The condensed nearest neighbor rule (1968).
    """

    def __init__(self, n_neighbors=1):
        self.n_neighbors = n_neighbors
        self.classifier = KNeighborsClassifier(n_neighbors=self.n_neighbors)
        self.sample_indices_ = []

    def select_data(self, X, y):
        # Validación de entrada: datos densos
        X, y = check_X_y(X, y, accept_sparse=False)

        idx_s = []  # índice de instancias seleccionadas
        classes = np.unique(y)
        self.classes_ = classes

        # Iniciar con una instancia aleatoria por clase
        for cur_class in classes:
            mask = np.where(y == cur_class)[0]
            choose_id = random.choice(mask)
            idx_s.append(choose_id)

        # Revisión aleatoria del resto de las instancias
        train_idx = list(range(len(y)))
        random.shuffle(train_idx)

        flag = True
        while flag:
            flag = False
            self.classifier.fit(X[idx_s], y[idx_s])
            for idx in train_idx:
                if idx not in idx_s:
                    x_i = X[idx].reshape(1, -1)
                    if self.classifier.predict(x_i) != [y[idx]]:
                        idx_s.append(idx)
                        self.classifier.fit(X[idx_s], y[idx_s])
                        flag = True

        self.X_ = X[idx_s]
        self.y_ = y[idx_s]
        self.sample_indices_ = sorted(idx_s)
        self.reduction_ = 1.0 - len(self.y_) / len(y)
        return self.X_, self.y_
```

```python


# Aplicar CNN
cnn = CNN(n_neighbors=1)
X_cnn, y_cnn = cnn.select_data(X, y)

# Visualización
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='k', alpha=0.7)
plt.title("Conjunto original")
plt.xlabel("Atributo 1")
plt.ylabel("Atributo 2")

plt.subplot(1, 2, 2)
plt.scatter(X_cnn[:, 0], X_cnn[:, 1], c=y_cnn, cmap='coolwarm', edgecolor='k', alpha=0.9)
plt.title("Conjunto condensado (CNN)")
plt.xlabel("Atributo 1")
plt.ylabel("Atributo 2")

plt.tight_layout()
plt.show()

print(f"Instancias originales: {len(y)}")
print(f"Instancias tras CNN:   {len(y_cnn)}")
print(f"Reducción: {100 * cnn.reduction_:.2f}%")
```

# Generación de Prototipos


## Objetivo general

Al igual que los métodos de **condensación de datos**, la generación de prototipos busca **reducir el conjunto de entrenamiento** a un número menor de instancias representativas (o **prototipos**). Sin embargo, va más allá:

> 🔍 **Los algoritmos de generación de prototipos no se limitan a seleccionar ejemplos existentes del conjunto original, sino que pueden también generar o ajustar nuevos puntos artificiales en el espacio de características.**

---

## ¿Qué diferencia a los métodos de generación de prototipos?

- No solo **eligen** ejemplos, sino que también pueden **modificarlos** o **crear nuevos**.
- Pueden **ajustar su ubicación** en el espacio de dimensión \( \mathbb{R}^d \).
- Utilizan **estrategias como agrupamiento, fusión iterativa o heurísticas espaciales** para lograr una representación compacta.

---

## Características deseables de los prototipos

- Deben **representar eficazmente** las distribuciones de clase.
- Deben **discriminar claramente** entre clases.
- El número total de prototipos debe ser **pequeño**, para reducir:
  - El **tiempo de cómputo** del clasificador.
  - El **espacio de almacenamiento**.

---

## Estrategias comunes para generar prototipos artificiales

### 1. Métodos basados en centroides
- Fusionan instancias similares de la **misma clase**.
- Los prototipos generados son los **centroides** de los grupos fusionados.
- Ejemplo típico: **Condensed LVQ, K-Means por clase**.

### 2. Métodos de ajuste de posicionamiento
- Ajustan iterativamente la **posición de los prototipos** usando reglas de actualización.
- Se basan en procedimientos de **optimización supervisada**.
- Ejemplo típico: **Learning Vector Quantization (LVQ)**.

### 3. Métodos de partición espacial
- Dividen el espacio de atributos en regiones mediante **heurísticas**.
- Seleccionan puntos representativos de cada región.
- Ejemplo típico: **SOFM, partición recursiva jerárquica**.

---

## Conclusión

> Los métodos de generación de prototipos ofrecen una forma flexible e inteligente de reducir conjuntos de datos, con la ventaja adicional de permitir el ajuste o creación de instancias que mejoran la generalización de los clasificadores.



# Práctica de Generación de Prototipos: K-Means

##¿Qué es K-medias? (K-means)

**K-medias** (K-means) es un algoritmo de **agrupamiento no supervisado** que busca dividir un conjunto de datos en **\( k \)** grupos (o clústeres) según su similitud. Es ampliamente utilizado en minería de datos, compresión de datos, generación de prototipos y segmentación.

---

### Objetivo

Agrupar instancias de forma que:

- Los puntos **dentro de un mismo clúster sean similares** (baja distancia entre ellos).
- Los puntos de **diferentes clústeres sean disímiles** (alta distancia entre grupos).

---

### ¿Cómo funciona el algoritmo?

1. **Inicialización**: seleccionar aleatoriamente \( k \) puntos como centroides.
2. **Asignación**: asignar cada punto al centroide más cercano.
3. **Actualización**: recalcular los centroides como la media de los puntos asignados.
4. **Iteración**: repetir los pasos 2 y 3 hasta convergencia.

---

### Función objetivo

El algoritmo intenta minimizar la **suma de las distancias cuadradas** entre los puntos y sus respectivos centroides:
\\[ \min \sum_{i=1}^{k} \sum_{x \in C_i} \left\| x - \mu_i \right\|^2\\]

Donde:

- \\( C_i \\): conjunto de puntos asignados al clúster \\( i \\)
- \\( \mu_i \\): centroide del clúster \\( i \\)


---

### Ventajas

- Simple y rápido.
- Eficiente en conjuntos grandes.
- Útil para compresión, visualización y preprocesamiento.

---

### ⚠️Limitaciones

- Se debe definir \( k \) (número de clústeres) de antemano.
- Puede converger a mínimos locales (sensible a la inicialización).
- No detecta bien clústeres de forma irregular o de tamaños muy distintos.

---

### 🧪 Ejemplos de aplicación

- Agrupamiento de clientes o usuarios.
- Reducción de datos mediante generación de prototipos.
- Compresión de imágenes mediante agrupamiento de colores.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Generar conjunto de datos sintético
X, y = make_classification(n_samples=300, n_features=2, n_informative=2,
                           n_redundant=0, n_clusters_per_class=1,
                           class_sep=1.0, random_state=42)

# 2. Escalar los datos para mejor desempeño de KMeans
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Aquí puedes modificar el número de instancias por clase
# 3. Número de prototipos por clase
prototypes_per_class = 10
classes = np.unique(y)

# 4. Aplicar KMeans por clase
prototypes = []
labels = []

for c in classes:
    # Subconjunto de la clase actual
    class_data = X_scaled[y == c]

    # KMeans sobre los datos de la clase
    kmeans = KMeans(n_clusters=prototypes_per_class, random_state=42)
    kmeans.fit(class_data)

    # Guardar prototipos y sus etiquetas
    prototypes.append(kmeans.cluster_centers_)
    labels.extend([c] * prototypes_per_class)

# 5. Concatenar resultados
X_proto = np.vstack(prototypes)
y_proto = np.array(labels)

# 6. Visualización
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Conjunto original
axes[0].scatter(X_scaled[:, 0], X_scaled[:, 1], c=y, cmap='coolwarm', edgecolor='k', alpha=0.7)
axes[0].set_title("Conjunto original (escalado)")
axes[0].set_xlabel("Atributo 1")
axes[0].set_ylabel("Atributo 2")

# Prototipos generados
axes[1].scatter(X_proto[:, 0], X_proto[:, 1], c=y_proto, cmap='coolwarm', edgecolor='k', s=100)
axes[1].set_title("Prototipos generados por K-Means")
axes[1].set_xlabel("Atributo 1")
axes[1].set_ylabel("Atributo 2")

plt.tight_layout()
plt.show()
```
